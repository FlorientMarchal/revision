# Démarrer Python avec un notebook avec Jupyter Notebook

## Tout d'abord observons l'interface de Jupyter Notebook
(Si vous utilisez un autre outil pour manipuler les notebooks, cherchez les équivalents de ce qui suit)
* La deuxième ligne propose des menus dont le menu **Help**. L'entrée **Notebook help** de ce menu est conseillée aux étudiants trouvant l'interface peu intuitive ou désirant explorer les possibilités.
* En troisième ligne de l'interface, se trouvent des boutons qui sont des raccourcis vers des opérations fréquentes (que nous découvrirons progressivement).

La page d'un notebook peut être vue comme un terminal de saisie élaboré. Elle est constituée de cellules. 
* Les boutons +, "ciseaux", copier, coller, déplacer vers le haut, déplacer vers le bas permettent de gérer ces cellules.
* Certaines cellules peuvent contenir du texte (au format Markdown (comme la cellule que vous lisez actuellement)) ou du code.
* Le langage "Markdown" peut être facilement retrouvé sur le web ou simplement dans le menu Help. L'édition (double-clic) de cellules dans ce format permet d'en avoir un aperçu.
* La liste déroulante au bout de la ligne de boutons permet de préciser le type de contenu d'une cellule (par défaut du code).

En mode édition, plusieurs méthodes existent pour valider le contenu d'une ligne de commande :
* SHIFT-ENTREE (ou bouton "Run" ou "Exécuter") pour exécuter un code (ou valider un texte "Markdown") (ajoute une ligne de commande si on exécute la dernière ligne de commande ; passe à la ligne de commande sinon)
* ALT-ENTREE pour exécuter et ajouter une ligne de commande (même si la cellule n'est pas la dernière)
* CTRL-ENTREE pour exécuter et rester sur la commande (en restant en mode commande exécutée et non en mode édition)

Remarques :
* en mode édition, ENTREE permet d'insérer une ligne (la saisie est multi-lignes)
* en mode édition, pour passer en mode commande sans exécuter : ESC
* en mode commande, pour passer en mode édition : ENTREE (ou clic de souris)

Si la page a été partiellement ou totalement exécutée, pour la réinitialiser, voir Kernel > Restart & Clear Output

### Affichage

*Remarques* 
* L'apprentissage des formats possibles doit venir par la pratique en cherchant à avoir des affichages clairs pour l'utilisateur et en consultant la documentation au fur et à mesure : par exemple, <a href="https://docs.python.org/fr/3/library/string.html#formatstrings">Python.org#formatage de chaîne</a>.
* Les possibilités de formatage ont évolué avec Python : voir par exemple https://realpython.com/python-string-formatting/.
* Il existe des modules spécialisés pour certains types d'affichage (voir librairie Pandas dans deuxième partie par exemple).
* Il existe d'autres manières de formater. 

```python
print("Bonjour")
```

```python
# exemples d'affectation multiple
a, b = 1/3, 2/3
```

```python
# exemple d'affichage - version 1
print(a)
print(b)
print(a+b)
```

```python
# exemple d'affichage - version 2 - Affichage plus parlant pour l'utiliseur
print("a =", a)
print("b =", b)
print("somme = ", a+b)
```

```python
# exemple d'affichage - version 3 - Affichage sur une ligne plus synthétique
print("a =", a, "\t\tb =", b, "\t\tsomme =", a+b) # \t = tabulation (\n = passage à la ligne)
```

```python
# exemple d'affichage - version 4 - approche plus spécifique Python
print("a = {:^}\t\tb = {}\t\tsomme = {}".format(a, b, a+b))
```

```python
# exemple d'affichage - version 5 - avec formatage (intérêt de la méthode format)
print("a = {:.3f} \t\tb = {:.3f} \t\tsomme = {:.3f}".format(a, b, a+b))
```

```python
# exemple d'affichage - version 6 - avec exemples de formatage
print("{:^7}|{:^7}|{:^7}".format("a","b","a+b"))# ^= centré ; 7 = largeur colonne en caractère
print("{:-<7}|{:-<7}|{:-<7}".format("","","")) # - = caractère de remplissage à répéter
print("{:^7.3f}|{:^7.3f}|{:^7.3f}".format(a, b,a+b)) # .3 = nombre de chiffres significatifs
```
