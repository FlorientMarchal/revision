# Codage par substitution
Dans cette partie, nous abordons un petit problème de cryptographie généralisant le code de César.
Dans ce codage de texte, chaque lettre est remplacée par une autre. L'ensemble des remplacements de lettres forme une permutation de l'alphabet de manière à permettre le décodage. Le codage sera stocké dans un dictionnaire (la clé est la lettre à remplacer, la valeur est la lettre de remplacement). Certaines lettres peuvent être remplacées par elles-même : cela ne sera pas nécessairement stocké dans le dictionnaire.

```python
# Exemple de dictionnaire de substitutions
substitutions = {"A":"B","B":"C","C":"A"}
```

Comme certaines lettres et certaines caractères peuvent ne pas être présents dans le dictionnaire, il nous faut écrire une fonction pour effectuer la substitution en renvoyant le caractère quand il n'apparaît pas dans le dictionnaire.

Complétez la fonction suivante.

```python
def substitute(lettre, dictionnaire_codage):
    """
    Retourne le codage de lettre si la lettre apparaît dans le dictionnaire de codage
    
    lettre : chaîne de longueur 1
    substitutions : dictionnaire de codage
    """
    if lettre in dictionnaire_codage:
        return dictionnaire_codage[lettre]
    else:
        return lettre
```

```python
# Tester ici votre fonction substitute
from aide_test import test
liste_tests = [
    (("A",substitutions),"B",""),
    (("B",substitutions),"C",""),
    (("C",substitutions),"A",""),
    (("D",substitutions),"D",""),
    ((".",substitutions),".","")
]
test( substitute, liste_tests)
```

À présent, écrivons la fonction qui va permettre de coder (ou décoder selon le dictionnaire qu'on utilisera) un texte.

*Remarque*. Vous pourrez :
* commencer par calculer la liste des caractères du résultat
* recréer une chaîne de caractères à l'aide de la méthode *join( )* de la classe *String*

```python
def codage_par_substitution(texte, dictionnaire_codage):
    """
    Encode le texte fournit en paramètre en utilisant un codage par substitutions
    utilisant le dictionnaire en paramètre
    
    texte : chaîne de caractères
    dictionnaire_codage : dictionnaire
    Retour : chaîne de caractères
    """
    return "".join([substitute(lettre,dictionnaire_codage) for lettre in texte])
```

```python
# Un test de la fonction de codage
print(codage_par_substitution("CARCBCDCARC!", substitutions))
```

## Codage par décalage
Nous nous intéressons à présent à la mise en place d'un dictionnaire pour un codage par décalage. Une valeur *d* de  décalage étant définie, les lettres sont remplacées par les lettres apparaissant *d* caractères plus loin dans l'alphabet géré circulairement. Par exemple, si *d* vaut 3, le 'Z' sera remplacé par 'C'. Dans ce qui suit, nous distinguons les caractères majuscules des caractères minuscules. Les autres caractères ne sont pas modifiés. 

Complétez la fonction suivante.

```python
def cree_dictionnaire_codage_par_décalage( d ):
    """
    Retourne un dictionnaire associant à chaque lettre, la lettre qui apparaît d lettres plus loin dans l'alphabet.
    Les caractères n'étant pas des lettres ne sont pas considérés.
    
    d : entier
    Précondition : -26 < d < 26 
    """
    maj_chars = [chr(i+ord('A')) for i in range(26)] # récupération de la liste des lettres majuscules
    min_chars = [chr(i+ord('a')) for i in range(26)] # récupération de la liste des lettres minuscules

    dico = {}
    for i, lettre in enumerate(maj_chars):
        dico[lettre] = maj_chars[(i+d) % len(maj_chars)]
    for i, lettre in enumerate(min_chars):
        dico[lettre] = min_chars[(i+d) % len(min_chars)]
    return dico
```

```python
# verification visuelle de la création du dictionnaire
print(cree_dictionnaire_codage_par_décalage( 3 ))
```
