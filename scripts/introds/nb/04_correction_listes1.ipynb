# Listes 1

Les listes jouent le rôle des tableaux dans d'autres langages. Mais la structure est beaucoup plus riche.

## Définition par extension
Ci-suivent deux exemples de définition de listes par extension.

```python
x = [] # La liste vide
```

```python
y = [2, 6, "rouge", True, 3.6] # Une liste avec des valeurs de types différents
```

L'exemple précédent montre que des éléments d'une liste peuvent être différents. Il reste conseillé dans de nombreuses situations de manipuler des listes d'éléments de même type (cela facilite, a minima, la compréhension).

# Afficher une liste
Le contenu d'une liste peut être affiché directement avec la fonction print.

*Astuce :* _print( *lis )_ permet d'afficher une liste *lis* sans les virgules séparatrices et sans les crochets délimitateurs de début et de fin de liste (en pratique la liste est transformé en type *Iterable* ce qui a pour effet que chaque élément de la liste est considéré comme un paramètre de la fonction *print( )*.

```python
# Affichez ici la liste y
print(y)
```

```python
# Affichez ici la liste y sans virgules ni crochets
print(*y)
```

```python
# consultez ici la document de la fonction print
help(print)
```

```python
# Affichez ici la liste y sans crochet, 
# avec des points-virgules pour séparer les éléments
# et avec un point d'exclamation 
# suivi d'un passage à la ligne à la fin de l'affichage
print(*y, sep=" ; ", end=" !\n") # Rem péda : des espaces non demandées ont été ajoutées 
                                # pour que ce soit plus agréable
```

## Longueur
La longueur d'une liste est son nombre d'éléments. Il s'obtient grâce à la fonction *len( )*.

```python
# Exemple
print("x contient {} éléments".format(len(x)))
print("y contient {} éléments".format(len(y)))
```

## Indices
Les éléments de la liste sont indicés à partir de 0. 
Si *i* est un indice d'une liste *lis*, *i* doit être un entier compris entre **0** et **len(lis)-1** (en gras le remplacement des ????? de l'énoncé).
L'élément à l'indice *i* pour une liste *lis* se note *lis[i]*.

```python
# affichez ici le premier élément de y
print(y[0])
```

```python
# affichez ci-dessous le dernier élément de y
# (faites-le en supposant que vous ne savez pas que la longueur de y est 5)
print(y[ len(y)-1])
```

## Parcourir tous les éléments d'une liste
Pour réaliser un traitement sur tous les éléments d'une liste, 
dans la plupart des langages, le principe consiste à faire varier
la valeur de l'indice grâce à une structure répétitive
(*while* ou *for*), avec une préférence pour un *for*
tant que tous les éléments de la liste sont systématiquement considérés.

Dans les questions suivantes, il est demandé
d'écrire une séquence d'instructions affichant tous
les éléments de la liste *y* chacun sur une ligne en respectant le format :<br/>
élément indice 0 = 2<br/>
élément indice 1 = 6<br/>
élément indice 2 = rouge<br/>
élément indice 3 = True<br/>
élément indice 4 = 3.6<br/>
sauf quand on n'a pas la valeur de l'indice, auquel cas, on affichera juste :<br/>
2<br/>
6<br/>
rouge<br/>
True<br/>
3.6

Vous devez maîtriser toutes ces techniques. Les deux premières sont des approches plus génériques (facilement transposables dans de nombreux langages). Les deux dernières sont plus pythonesques.
Pour toutes les questions, répondez-y en supposant que vous ne savez pas que la longueur de y est 5.


```python
# Affichez ici tous les éléments de la liste y les uns en dessous des autres
# Utilisez un while
i = 0
while i < len(y):
    print("élément d'indice {} = {}".format(i, y[i]))
    i += 1
```

```python
# Affichez ici tous les éléments de la liste y les uns en dessous des autres
# Utilisez un for avec range
for i in range(len(y)): # ou for i in range(0, len(y),1):
    print("élément d'indice {} = {}".format(i, y[i]))
```

```python
# Affichez ici tous les éléments de la liste y les uns en dessous des autres
# Utilisez un for elt in liste
for elt in y:
    print(elt)
```

```python
# Affichez ici tous les éléments de la liste y les uns en dessous des autres
# Utilisez un for avec enumerate
for i, elt in enumerate(y):
    print("élément d'indice {} = {}".format(i, elt))
```

## Un peu plus sur les indices
Python accepte des indices négatifs. Si *lis* est une liste, *lis[-1]* est le dernier élément, *lis[-2]* est l'avant dernier, *lis[-len(lis)]* est le premier.

Il peut être utile d'avoir cette particularité en tête car elle peut être à l'origine d'erreur (parfois difficile à trouver)

```python
# Afficher le dernier élément de y en utilisant un indice négatif
print(y[-1])
```

```python
# Afficher le premier élément de y en utilisant un indice négatif
# (faites-le en supposant que vous ne savez pas que la longueur de y est 5)
print(y[-len(y)])
```

```python
# Faites varier une variable i de -1 à -len(y) à l'aide d'un for avec range
# et afficher les éléments de y aux indices i.
# (Observez que l'affichage se fait en sens inverse)
for i in range(-1,-len(y)-1,-1):
    print("élement en position {} vaut {}".format(i,y[i]))
```

L'approche précédente est une approche très générique. En python, pour traiter une liste en ordre inverse, la fonction *reversed( )* sera utilisée (et la fonction *enumerate( )*). Le résultat de cette fonction *reversed( )* est un itérateur et non une liste (voir https://docs.python.org/3/library/functions.html#reversed).

```python
# Reproduisez l'affichage précédent en utilisant reversed et enumerate
for i, elt in enumerate(reversed(y)):
    print("élement en position {} vaut {}".format(i,y[i]))
```

```python
# Reproduisez l'affichage précédent en utilisant reversed et enumerate
for i, elt in reversed(list(enumerate(y))):
    print("élement en position {} vaut {}".format(-i-1,elt))
```
