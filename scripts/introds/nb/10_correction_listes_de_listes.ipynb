## Extraction de ligne

```python
def extraire_ligne(t2D, i):
    """extrait la ligne i
    
    t2D : listes de listes
    Précondition : 0 <= i < len(t2D)
    """
    return t2D[i] 
```

```python
# Test extraction ligne
print(extraire_ligne(tab_test, 2))
```

## Extraction de colonne

```python
def extraire_colonne(t2D, j):
    """ extrait la colonne j du tableau 2D t2D
    
    t2D : listes de listes
    Précondition : les lignes de t2D ont toutes une longueur strictement supérieure à j
    """
    return [t2D[i][j] for i in range(len(t2D))]
```

```python
# Test extraction colonne
print(extraire_colonne(tab_test, 2))
```

## Extraction de bloc

```python
def extraire_neuvieme(t2D, i, j):
    """ 
    extraction du neuvième du tableau contenant la cellule i, j et retour sous forme d'une liste
    
    t2D : une liste de len(tD2) valeurs
    Précondition : len(t2D) est un multiple de 3 
    i, j : entier
    Précondition : 0 <= i, j < len(t2D)
    """
    longueur_tiers = len(t2D)//3 # Calcul du tiers de la longueur
    ligne_neuvieme = i // longueur_tiers # calcul de l'indice (0, 1 ou 2) de ligne du neuvième à récupérer
    col_neuvieme = j // longueur_tiers # calcul de l'indice (0, 1 ou 2) de colonne du neuvième à récupérer
    ind_ligne_debut = ligne_neuvieme * longueur_tiers # indice de la première ligne de bloc à récupérer
    ind_ligne_apres_fin = (ligne_neuvieme+1) * longueur_tiers # indice de ligne suivant la dernière ligne 
                                                                # du bloc à récupérer
    ind_col_debut = col_neuvieme * longueur_tiers # indice de la première colonne de bloc à récupérer
    ind_col_apres_fin = (col_neuvieme+1) * longueur_tiers # indice de ligne suivant la dernière colonne 
                                                                # du bloc à récupérer
    # Ci-après récupération, sous forme d'une liste, du bloc de valeurs constitué
    # des éléments d'indice de ligne >= ind_ligne_debut et < ind_ligne_apres_fin
    # et d'indice de colonne >= ind_col_debut et < ind_col_apres_fin
    return [t2D[i][j] \
            for i in range(ind_ligne_debut, ind_ligne_apres_fin, 1) \
            for j in range(ind_col_debut, ind_col_apres_fin, 1)]
```

```python
# Test extraction neuvième
print(extraire_neuvieme(tab_test, 4, 7))
```

## Test liste complète
Le rôle de la fonction suivante se comprend dans le contexte de la résolution d'un Sudoku.

```python
def tous_les_chiffres( lis ):
    """ 
    Vérifie que lis contient tous les chiffres de 1 à 9 
    et uniquement ces chiffres
    
    lis : liste d'entiers
    """
    return set(lis) == set([i for i in range(1,10)])
```

## Vérification Sudoku
La fonction de vérification qu'une grille est une solution d'un Sudoku :
* toutes les cases sont remplies (on suppose que la grille contient que des chiffres ou la valeur None qui désigne une case non remplie
* chacune des lignes contient exactement tous les chiffres de 1 à 9
* chacune des colonnes contient exactement tous les chiffres de 1 à 9
* chacun des neuvièmes contient exactement tous les chiffres de 1 à 9

Prenez le temps de comprendre l'unique ligne d'instruction de cette fonction.

```python
def verif_sudoku( t2D ):
    """ verifie qu'une grille est une solution Sudoku """
    return all([t2D[i][j]!=None for i in range(9) for j in range(9)]) and \
            all([tous_les_chiffres(extraire_ligne(t2D,i)) for i in range(9)]) and \
            all([tous_les_chiffres(extraire_colonne(t2D,i)) for i in range(9)]) and \
            all([tous_les_chiffres(extraire_neuvieme(t2D,i,j)) \
                 for i in range(0,9,3) for j in range(0,9,3)])
```

```python
# Test de la fonction de vérification - n'hésitez pas à en prendre d'autres
sudoku = [
    [ 1, 2, 3, 4, 5, 6, 7, 8, 9],
    [ 4, 5, 6, 7, 8, 9, 1, 2, 3],
    [ 7, 8, 9, 1, 2, 3, 4, 5, 6],
    [ 2, 3, 4, 5, 6, 7, 8, 9, 1],
    [ 5, 6, 7, 8, 9, 1, 2, 3, 4],
    [ 8, 9, 1, 2, 3, 4, 5, 6, 7],
    [ 3, 4, 5, 6, 7, 8, 9, 1, 2],
    [ 6, 7, 8, 9, 1, 2, 3, 4, 5],
    [ 9, 1, 2, 3, 4, 5, 6, 7, 8]
]
print(verif_sudoku(sudoku))
```

## Affichage
La fonction suivante a pour but d'afficher une grille de Sudoku de manière plus proche de ce que peut s'attendre un utilisateur (la fonction peut sans doute être améliorée ; la fonction peut être utilisée dans bien d'autres contextes

```python
def affiche_2D(t2D):
    """
    Affiche une liste de listes sans les délimitateurs de liste
    """
    for i in range(len(t2D)):
        print(*t2D[i])
```

```python
def affiche_2D(t2D):
    """
    Affiche une liste de listes sans les délimitateurs de liste
    """
    for i in range(len(t2D)):
        a_afficher = []
        for j in range(len(t2D[i])):
            if t2D[i][j] == None: 
                a_afficher.append(" ")
            else: 
                a_afficher.append(t2D[i][j])
        print(*a_afficher)
```

## Résolution de Sudoku
Là encore, la fonction de résolution de Sudoku est fournie. Cherchez à la comprendre. La fonction est récursive : son principe de base consiste à essayer de remplir les trous avec les chiffres un à un si le chiffre n'apparaît et passer au remplissage du trou suivant ; un retour en arrière dans le remplissage est réalisé pour tester un autre chiffre. Ce n'est certainement pas comme cela qu'un humain résolverait une grille mais la puissance des ordinateurs le permet (malgré une combinatoire a priori défavorable).

On pourra observer le paramètre facultatif *niveau*.

```python
def resoudre_sudoku(t2D, niveau = 0):
    """
    Affiche toutes les solutions d'une grille de Sudoku.
    
    t2D : grille 9x9 d'entiers entre 1 et 9 et valeur None (case non remplie)
    niveau indique la prochaine la prochaine case à essayer de remplir 
    (si elle ne contient pas déjà un chiffre)
    (les cases sont numérotées de 1 à 81 de la gauche vers la droite et du haut vers le bas)
    """
    if niveau == 81:
        affiche_2D(t2D)
    else:
        i, j = niveau % 9, niveau // 9 # Calcul des indices de la case à étudier
        if t2D[i][j] == None: # case non remplie
            ligne = extraire_ligne(t2D, i)
            colonne = extraire_colonne(t2D, j)
            neuvieme = extraire_neuvieme(t2D, i, j)
            for elt in range(1, 10): # test de tous les chiffres les uns après les autres
                if (elt not in ligne) and (elt not in colonne) and (elt not in neuvieme):
                    t2D[i][j] = elt
                    resoudre_sudoku(t2D, niveau+1)
                    t2D[i][j] = None
        else:
            resoudre_sudoku(t2D, niveau+1)
```

```python
# Un exemple de grille à remplir
sudoku = [
    [ 1, 2, None, 4, 5, None, 7, 8, 9],
    [ 4, 5, 6, 7, 8, 9, 1, None, 3],
    [ None, 8, 9, 1, 2, 3, 4, 5, 6],
    [ 2, None, 4, 5, 6, 7, 8, 9, 1],
    [ 5, 6, 7, 8, None, 1, 2, None, 4],
    [ 8, 9, None, 2, 3, None, 5, 6, 7],
    [ 3, 4, 5, 6, 7, 8, 9, None, 2],
    [ 6, 7, 8, 9, None, 2, 3, 4, 5],
    [ 9, 1, 2, 3, 4, 5, 6, 7, None]
]
affiche_2D(sudoku)
```

```python
resoudre_sudoku(sudoku)
```
