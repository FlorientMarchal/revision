# Définir de nouveaux types
Pour définir un nouveau type, il faut utiliser la possibilité de faire de la programmation Objet avec Python.

## Type Enum
Définir une classe héritée du type Enum permet de définir un nombre fini de valeurs. L'intérêt principal est de fixer des constantes (on ne peut pas changer les valeurs après définition) qui pourront être utilisées sans ambigüité dans la suite.

```python
# Exemple
from enum import Enum
class Etat(Enum):
    LIBRE = 1
    EMPRUNTE = 2
```

```python
liste_etats= [Etat.LIBRE, Etat.EMPRUNTE, Etat.EMPRUNTE, Etat.EMPRUNTE, Etat.LIBRE]
```

```python
1 in liste_etats
```

```python
Etat.LIBRE in liste_etats
```

```python
Etat.LIBRE = 5
```

## Définir un type d'enregistrement
La notion d'enregistrement précède la notion d'objet. Il s'agit globalement de regrouper en un lieu un ensemble d'attributs modélisant un "objet" du réel. En Python, il suffit de définir une classe sans méthode (en pratique, on utilisera la puissance de l'approche et des méthodes seront très certainement développées).

Avec la notion de dictionnaire, on peut parfois se passer de définir formellement une classe d'objets mais il faut alors être très prudent à bien nommer les clés des attributs toujours de la même manière. Définir une classe d'objets permet de bénéficier de la vérifiation syntaxique des codes écrits. 

Nous donnons ci-après, sans le développer, un exemmple de liste de livres définie sous forme d'une liste de dictionnaire.

```python
# exemple avec dictionnaire
biblio = [ { "numL":"numL01", "titre":"Titre 01", "auteur":"Auteur1","etat":Etat.LIBRE},
         { "numL":"numL02", "titre":"Titre 02", "auteur":"Auteur2","etat":Etat.EMPRUNTE}]
print(biblio)
```

Nous rappelons à présent comment un type *Livre* peut être défini et utilisé sous forme d'une classe.

```python
class Livre:
    def __init__( self, numL, titre, auteur, etat):
        self.numL = numL
        self.titre = titre
        self.auteur = auteur
        self.etat = etat
        
    def __str__( self ):
        """ Méthode retournant une description textuelle d'un livre.
            Cette description est utilisée par la fonction print
            """
        return "Numéro = {}, Titre = {}, Auteur = {}, Etat = {}".format(self.numL,
                                                                        self.titre,
                                                                        self.auteur,
                                                                        self.etat)
```

```python
# Création d'une instance
Livre1 = Livre("numL01","Titre 01","Auteur1",Etat.LIBRE)
print(Livre1)  # Voyez l'effet de méthode __str__( )
```

```python
# Création d'une liste de 2 instances et affichage
Livre2 = Livre("numL02","Titre 02","Auteur2",Etat.LIBRE)
biblio = [ Livre1, Livre2]
print(biblio) # Commprenez vous ce que donne cet affichage ?
```

Comme l'affichage d'une liste de valeur de type Livre ne s'affiche pas naturellement correctement, définisssons une fonction d'affichage. 

Complétez la fonction suivante.

```python
def affichage_biblio( lis_livres):
    """ permet d'afficher une liste de livres """
    for livre in lis_livres:
        print(livre)
```

```python
affichage_biblio(biblio)
```

Afin de pouvoir définir une liste de livres plus longue et plus facilement, nous utilisons un fichier CSV et une fonction de lecture de fichier CSV définie dans le module *manipulation_csv*. Cette fonction retourne une liste de listes contenant le contenu du fichier CSV. (Vous verrez avec la bibliothèque *Panda* une autre approche, plus riche, pour lire des fichiers de données).

La fonction *listes2D_vers_liste_Livre* permet de transformer le résultat de cette lecture en une liste de valeurs de type *Livre*.

```python
def etat( val ):
    """ Retourne l'état correspondant à val """
    if val == "1":
        return Etat.LIBRE
    elif val == "2":
        return Etat.EMPRUNTE
    
def listes2D_vers_liste_Livre( lis2D ):
    """
    Transforme la liste lis2D en une liste de Livre
    
    Donnée : lis2D : liste de listes de chaîne de caractères
    Précondition : toutes les listes de lis2D ont 4 éléments dans l'ordre
        (numéro, titre, auteur, etat)
    Résultat : liste de valeurs de type Livre
    """
    return [ Livre(lis[0], lis[1], lis[2], etat(lis[3])) for lis in lis2D]
```

```python
from manipulation_csv import lire_fichier_csv
biblio = listes2D_vers_liste_Livre( lire_fichier_csv("bibliotheque.csv"))
affichage_biblio(biblio)
```

Trions la liste pour y voir plus clair (si le critère de tri ne vous convient pas, changez-le).

```python
from operator import attrgetter
biblio = sorted( biblio, key=attrgetter("titre","numL"))
```

```python
affichage_biblio(biblio)
```

**Exercice.** Complétez la fonction suivante et testez-la. Vous pouvez réfléchir à une approche pythonesque et à une approche plus générique.

```python
# Approche python
def nb_livres_libres( biblio ):
    """ Retourne le nombre de livres libres dans biblio
    
    biblio : liste de valeurs de type Livre
    """
    return sum([ 1 for livre in biblio if livre.etat == Etat.LIBRE])
```

```python
# Approche générique
def nb_livres_libres( biblio ):
    """ Retourne le nombre de livres libres dans biblio
    
    biblio : liste de valeurs de type Livre
    """
    nb = 0
    for i in range(0, len(biblio), 1):
        if biblio[i].etat == Etat.LIBRE:
            nb += 1
    return nb
```
