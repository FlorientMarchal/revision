## Premier exemple

```python
def saisie_liste( n ):
    """
    Saisie et retourne une liste de n valeurs
    
    Donnée : n est un entier
    Résultat (en retour de fonction) : une liste de n valeurs
    Remarque : si n <= 0, la liste vide est retournée
    """
    print("Donnez {} valeurs".format(n))
    lis = []
    for i in range(0, n, 1):
        val = input(str((i+1))+"ème valeur : ")
        lis.append( val )
    return lis
```

## Somme
Une fonction très pratique sur les listes est la fonction *sum()* qui permet de calculer la somme des éléments d'une liste (et d'autres structures itérables comme les ensembles). Utilisez cette fonction pour afficher pour chacun des éléments de la liste *liste_listes*, l'élément suivi de la somme de ces constituants.

```python
lis1 = [1, 2, 3, 4]
lis2 = [3.5, 0.5, 6.0]
lis3 = [1, 9.0]
lis4 = [2, 3, True, "rouge", 2.5, 2.5]
liste_listes = [lis1, lis2, lis3, [], lis4]

# écrivez ci-après l'affichage des éléments de liste_listes
for lis in liste_listes:
    print(lis, ":", sum(lis))
# Remarque pédagogique : l'erreur qui apparaît est normale !
```

**Exercice.** complétez la fonction suivante qui doit produire le même résultat que la fonction *sum( )*. Le corps de la fonction sera écrite de manière générique.

*Remarque de correction*. Écrire de manière générique c'est écrire de manière la plus passe-partout, c'est-à-dire, pouvant s'adapter très rapidement à la plupart des langages. Comme de nombreux langages ne prévoient pas l'utilisation d'une struture de contrôle du type *for élément in liste*, il ne faut pas l'écrire dans une approche générique.

```python
def somme( lis ):
    """
    Retourne la somme des éléments de lis
    
    lis : listes de nombres
    """
    s = 0
    for i in range(len(lis)):
        s += lis[i]
    return s
```

## Deuxième exemple + rappel technique du drapeau

Pour écrire la version suivante, nous utilisons la technique du drapeau permettant de sortir d'une boucle quand une variable (jouant le rôle d'un drapeau) change de valeur (levée du drapeau). Cette technique de base est à connaître.

```python
def recherche( a_chercher, lis):
    """
    Retourne le plus petit indice de la valeur a_chercher dans lis. Retourne -1 en cas d'absence.
    
    Donnée : a_chercher est une valeur de type quelconque
    Donnée : lis est une liste de valeurs de types compatibles avec la valeur cherchée
    Résultat en retour de fonction : indice de a_chercher dans lis si val y
        est présent, -1 sinon
    """
    i = 0
    indice = -1
    while i < len(lis) and indice == -1:
        if lis[i] == a_chercher:
            indice = i
        else:
            i += 1
    return indice
```

```python
liste_test = [1, "rouge", 3.2, "2"]
valeurs_cherchees = [1, "rouge", 3.2, "2", 3, 2, "bleu"]
for elt in valeurs_cherchees:
    print("À quel indice est présente la valeur {} dans la liste ? Réponse = {}".format(elt,
                                                                                    recherche(elt,liste_test)))
```

## Usage du return
La syntaxe de l'instruction *return* varie d'un langage à l'autre. Dans Python (mais aussi C, C++, java, ...), l'instruction *return* arrête l'exécution d'une fonction et renvoie immédiatement la valeur de l'expression après le mot return. Cela peut permettre de simplifier l'écriture d'une fonction comme montré dans les deux blocs suivants. Cela étant dans d'autres langages (Pascal, Ada, ...), l'instruction *return* n'arrête pas l'exécution de la fonction mais stocke la valeur dans une zone mémoire pour la renvoyer quand la fonction est terminée.

Pour pouvoir reprendre des codes et pour en faciliter la compréhension, il est souvent préférable de **ne pas utiliser cette instruction *return* dans le corps d'une boucle**, et d'écrire des fonctions n'ayant qu'un seul return à la fin. Comme vous pouvez être amenés à rencontrer ce type d'algorithme, nous vous en donnons des exemples même si nous déconseillons cette approche souvent peu lisible.

```python
def recherche( a_chercher, lis):
    """
    Retourne le plus petit indice de la valeur a_chercher dans lis. Retourne -1 en cas d'absence.
    
    Donnée : a_chercher est une valeur de type quelconque
    Donnée : lis est une liste de valeurs de types compatibles avec la valeur cherchée
    Résultat en retour de fonction : indice de a_chercher dans lis si val y
        est présent, -1 sinon
    """
    i = 0
    while i < len(lis):
        if lis[i] == a_chercher:
            return i
        i += 1
    return -1
```

```python
liste_test = [1, "rouge", 3.2, "2"]
valeurs_cherchees = [1, "rouge", 3.2, "2", 3, 2, "bleu"]
for elt in valeurs_cherchees:
    print("À quel indice est présente la valeur {} dans la liste ? Réponse = {}".format(elt,
                                                                                    recherche(elt,liste_test)))
```

Comme le montre l'exemple précédent, en cas d'absence de la valeur de la liste, la méthode *index()* provoque une exception. Cette exception peut être traitée et capturée de la manière suivante.

```python
liste_test = [1, "rouge", 3.2, "2"]
valeurs_cherchees = [1, "rouge", 3.2, "2", 3, 2, "bleu"]
for elt in valeurs_cherchees:
    try:
        print("À quel indice est présente la valeur {} dans la liste ? Réponse = {}".format(elt,
                                                                                    liste_test.index(elt)))
    except:
        print("La valeur {} est absente de la liste".format(elt))
```

Réécrire le code de la fonction *recherche* en utilisant la méthode *index()*.

```python
def recherche( a_chercher, lis):
    """
    Retourne le plus petit indice de la valeur a_chercher dans lis. Retourne -1 en cas d'absence.
    
    Donnée : a_chercher est une valeur de type quelconque
    Donnée : lis est une liste de valeurs de types compatibles avec la valeur cherchée
    Résultat en retour de fonction : indice de a_chercher dans lis si val y
        est présent, -1 sinon
    """
    try:
        return lis.index(a_chercher)
    except:
        return -1
```

## La fonction *any( )*
Dans certains cas, en Python, face à des problèmes un tout petit peu plus complexe que la simple recherche de présence d'éléments, on rencontre une autre approche. Prenons comme exemple le problème consistant à déterminer s'il existe un nombre pair dans une liste d'entiers. Ci-suit la démarche souvent rencontrée.

```python
lis = [ 3, 5, 1, -1, 0, 2, 9] # Liste de travail
# Calcul d'une liste de booléen indiquant pour chaque élément s'il est pair
# La liste ci-dessous est construite par compréhension : nous reviendrons sur ce type de construction
print(lis)
elts_pairs = [ elt % 2 == 0 for elt in lis]
print(elts_pairs)
existe_pair = any(elts_pairs) # any est vrai si un des éléments de la liste est vrai
print("Existe-t-il dans la liste {} une valeur paire ?\nRéponse : {}".format(lis, existe_pair))

# N'hésitez pas à réaliser d'autres tests en modifiant lis
# Pour comprendre la séquence, peut-être aurez-vous besoin d'afficher le contenu des variables intermédiaires.
```

Remarque. Cette approche a une complexité linéaire en le nombre d'éléments de la liste initiale mais on pourrait regretter, par rapport à une approche générique utilisant la technique du drapeau (que vous devez savoir utiliser dans ce cas), le fait que le traitement ne s'arrête pas dès qu'on rencontre un élément. En pratique pour des petites listes, la différence de gain de temps est minime entre les diverses approches. Pour des listes un peu plus  grandes, le fait d'utiliser une fonction native Python peut permettre d'avoir un traitement plus rapide que l'approche générique (qui globalement ne divise le temps d'exécution en moyenne que par 2, facteur qui peut être récupéré par le fait que les fonctions natives sont optimisées).

## Test d'universalité, fonction *all( )*
Une autre classe de problème sur les listes est celle des problèmes du type : "est-ce que toutes les valeurs de la liste vérifie telle propriété ?". Par exemple, on peut considérer le problème : est-ce que toutes les valeurs de la liste sont pairs ?

La fonction *all( )* permet de traiter ce type de problème avec l'approche précédente.

```python
def my_all( lis_booleens ):
    """
    Retourne la valeur True si tous les éléments de la liste 
        valent True, retourne False sinon
    (implémente le "et logique" sur toutes les valeurs 
        de lis_booleens)
    
    Données :
        lis_booleens : liste de booléens
    Type résultat : booléen
    """
    ok = True
    i = 0
    while i < len(lis_booleens) and ok:
        ok = lis_booleens[i]
        i+=1
    return ok
```

```python
# Deuxième test avec résultat positif 
lis = [ 32, 52, 12, -12, 0, 2, 92]
elts_pairs = [ elt % 2 == 0 for elt in lis] 
tous_pair = my_all(elts_pairs) # any est vrai si un des éléments de la liste est vrai
print("Est-ce que toutes les valeurs de la liste {} sont paires ?\nRéponse : {}".format(lis, tous_pair))

lis = [ 3, 5, 1, -1, 0, 2, 9]
elts_pairs = [ elt % 2 == 0 for elt in lis] 
tous_pair = my_all(elts_pairs) # any est vrai si un des éléments de la liste est vrai
print("Est-ce que toutes les valeurs de la liste {} sont paires ?\nRéponse : {}".format(lis, tous_pair))
```

## None
Certaines fonctions (affichage, enregistrement dans un fichier) n'ont pas de valeur de retour. En Python, lors de l'exécution d'une fonction sans return, la valeur *None* est retournée.

Il est conseillé de n'écrire que des fonctions dont le comportement est homogène : retour de valeurs dans tous les cas ou jamais, même type pour le retour. Dans le cas où vous auriez oublié le traitement de certains cas lors de l'écriture d'une fonction, la valeur None peut apparaître. Il est bon de le connaître pour déboguer un code.
