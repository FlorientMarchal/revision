## Concaténation et extension
L'opérateur de concaténation (+) permet d'obtenir une liste constituée des éléments d'une première liste suivi des éléments d'une seconde. Cette opération à une complexité en temps en O de la somme des longueurs des deux listes concaténées.

```python
# Exemple
[1, 2] + [6, 9, 10]
```

Il est possible d'utiliser la méthode extend( ) de list. Elle place tous les éléments de la liste paramètre à la suite de la liste modifiée. Cette méthode a une complexité en temps en O du nombre d'éléments de la liste paramètre utilisée lors de l'appel de la méthode.

```python
lis1 = [1, 2]
lis2 = [6, 9, 10]
# Sur la ligne suivante, modifiez lis1 pour lui ajouter les éléments de lis2
lis1.extend(lis2)
# Vérifiez ci-dessous que lis1 a bien été modifié (et pas lis2)
print("lis1 = ", lis1)
print("lis2 = ", lis2)
```

À la place de extend( ), on pourra utiliser l'instruction +=.

```python
# Réécrivez le code précédent en utilisant += à la place d'extend
lis1 = [1, 2]
lis2 = [6, 9, 10]
# Sur la ligne suivante, modifiez lis1 pour lui ajouter les éléments de lis2
lis1 += lis2
# Vérifiez ci-dessous que lis1 a bien été modifié (et pas lis2)
print("lis1 = ", lis1)
print("lis2 = ", lis2)
```

Si on ne veut ajouter qu'un élément, on pourra utiliser la méthode append( ) qui s'exécute en un temps en O(1).

```python
lis = [3, 8, 10, -1, 2]
# Ajoutez la valeur 100
lis.append(100)
# Vérification
print(lis)
```

Ci-suivent d'autres exemples d'insertions de valeur.

```python
lis = [300]+lis # insertion en tête - complexité en O de len(lis)
# Il existe d'autres méthodes non abordées ici
print(lis)
lis.insert(2, -5)
print(lis)
```

**Exercice 1.** Complétez la fonction suivante.

```python
def carres( lis ):
    """ 
    Retourne la liste des carrés des valeurs de lis
    
    Données : lis : liste de nombre
    """
    res = []
    for elt in lis:
        res.append(elt*elt)
    return res
# Remarque pédagogique : l'exercice ne donne pas de contrainte, 
# les étudiants peuvent l'écrire avec la structure de contrôle de leur choix
# La réponse ici est donnée avec la structure de contrôle adaptée selon les
# règles de choix en programmation Python (mais les listes par compréhension
# fournissent une solution encore plus pythonesque)
```

## Listes par compréhension

Définir une liste par compréhension consiste à la décrire. La possibilité de construire une liste par compréhension est une spécificité Python. Cela mène à des codes très concis tout en restant clairs. Ci-suit un exemple

```python
lis = [1, 3, 0.4, 2, 0.]
squares = [lis[i]**2 for i in range(len(lis))]
print(squares)
```

Vous aurez observé que l'exemple précédent fournit une autre manière de coder le corps de la fonction carrés précédemment vue (en une ligne). Vous devez savoir très rapidement construire une liste aussi bien par le mécanisme de liste par compréhension qu'avec une approche plus générique comme vous l'avez potentiellement réalisé dans l'exercice précédent.

```python
# Remarque de correction
# Une autre initialisation par extension
squares = [elt**2 for elt in lis]
print(squares)
```

**Exercice 2.** Que fait la fonction suivante ? Complétez-la.

```python
def mystere1( n ):
    """
    Retourne la liste des entiers impairs strictement inférieurs à n
    
    Donnée : n entier 
    Précondition n >= 0
    """
    lis = []
    for i in range(1,n,2):
        lis.append(i)
    return lis
```

```python
# Exemple d'exécution (n'hésitez pas à prendre d'autres exemmples)
print(mystere1(21))
```

**Question 5.** Est-ce que la séquence suivante réalise le même traitement que la fonction précédente ?

```python
def mystere2( n ):
    """
    Retourne la liste des n premiers entiers impairs
    Donnée : n entier
    
    Précondition n >= 0
    """
    lis = []
    for i in range(n):
        lis.append(2*i+1)
    return lis
```

```python
# Exemple d'exécution
print(mystere2(5))
```

**Exercice 6.** Réécrivez ci-dessous la fonction précédente de manière à ce que son corps ne soit constitué que d'une seule ligne.

```python
def mystere2bis( n ):
    """
    Retourne la liste des n premiers entiers impairs
    """
    return [2*i+1 for i in range(n)]
```
