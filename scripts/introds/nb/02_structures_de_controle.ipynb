## Un exemple de structure conditionnelle

```python
# Comprenez et tester le code suivant

x = input("Quel est votre nom ? ")
if len(x) == 0: # ici len = longueur d'une chaîne de caractères
    print("Erreur : vous n'avez rien saisi")
else:
    print("Bonjour " + x) # Le + est l'opérateur de concaténation de chaînes de caractères
```

Remarque : pour tester le code précédent, il faut réaliser au moins 2 exécutions :
* Une avec votre nom (ou n'importe quelle suite de caractères non vide)
* Une en ne saisissant aucun caractère

```python
# Comprenez et tester le code suivant

x = input("Donnez un nombre entre 0 et 9 : ")
if len(x) == 0:
    print("Erreur vous n'avez rien saisi")
else:
    if len(x) >=2 :
        print("Erreur : vous avez saisi trop d'information")
    else:
        if x.isdigit(): # https://docs.python.org/3/library/stdtypes.html#str.isdigit
            c = 10 - int(x) # int : transformation de chaîne en entier
            print("Le complément à 10 de votre nombre est", c)
        else:
            print("Erreur : vous n'avez pas saisi un chiffre")
```

Avez-vous bien fait au moins 4 tests ? (un pour chacun des 4 cas possibles)

```python
# Comprenez et tester le code suivant

x = input("Donnez un caractère : ")
if len(x) == 0: # == pour tester égalité
    print("Erreur vous n'avez rien saisi")
elif len(x) >= 2:
    print("Erreur vous avez saisi trop de caractères")
else:
    if x == "a":
        print("Oh! La première lettre de l'alphabet")
    if x != "z": # != l'opérateur pour tester la différence 
        print("Votre lettre n'est pas la dernière lettre de l'alphabet")
    if x in ["a","e","i","o","u","y"]: # in : appartenance d'un élément dans une liste
        print("Votre lettre est une voyelle")
    else:
        print("Votre lettre est une consonne en minuscule ou une majuscule ou n'est pas une lettre")
```

## Un exemple d'utilisation de la structure de contrôle while

```python
# Comprenez et tester le code suivant

i = 1
while i <= 10: # Faire plusieurs essais avec 100, 1000 - pour voir l'évolution de l'interface
    print(i)
    i += 1   # ou i = i + 1 
```

## Le même exemple avec la structure de contrôle for

```python
# Comprenez et tester le code suivant

for i in range(1, 11, 1): # Faire plusieurs essais avec 101, 1001 - pour voir l'évolution de l'interface
    print(i)
```

```python
# Comprenez et tester le code suivant

for i in range(10): # Faire plusieurs essais avec 100, 1000
    print(i)
```

*range()* est une fonction renvoyant un itérateur (plus ou moins une liste mais avec des méthodes (au sens programmation objet)) permettant son parcours.

**Syntaxe simplifiée**

*range([valeur_initiale],seuil, [pas])*
* Par défaut la *valeur initiale* vaut 0 et le pas 1
* *seuil* est une valeur d'arrêt (non prise)
* Il faut préciser la *valeur initiale*, si on veut préciser le *pas*
* Le *pas* peut être négatif

## Variation Python

Cette variante Python n'existe pas dans tous les langages. Vous devez savoir vous en passer dans une approche générique.

```python
liste_fruits = ["poire", "pomme", "banane"]
```

```python
# Comprenez et tester le code suivant

# for elt in liste -> utile quand on n'a pas besoin d'utiliser les indices
for fruit in liste_fruits:
    print("une {} est un fruit".format(fruit))
```

Equivalent avec un parcours sur indices

```python
# Comprenez et tester le code suivant

# Avec un parcours d'indice plus classique (pour illustration)
for i in range( 0, len(liste_fruits), 1):
    print("#{} : une {} est un fruit".format((i+1),liste_fruits[i]))
```

```python
# Comprenez et tester le code suivant

# Approche plus pythonesque quand on a besoin des indices
for i, fruit in enumerate(liste_fruits):
    print("#{} : une {} est un fruit".format((i+1),liste_fruits[i]))
```

## Comment choisir sa structure de contrôle
* Toujours chercher à utiliser la structure de contrôle qui explique le mieux votre algorithme.
* En approche passe partout ou générique (reproductible dans un langage ne comportant pas les spécificités de Python)
    * Préférez *for* pour exprimer un parcours complet de votre structure de données pour montrer le côté systématique de l'algorithme.
    * Utilisez *while* si l'algorithme peut s'arrêter avant la fin d'un parcours (éviter les *break*, *continue* et *return* dans les boucles : leur analyse est rapidement complexe et le débogage peut s'avérer complexe).
* En approche Python
    * Utilisez le *for elt in liste* dès que possible
    * Utilisez *enumerate* si vous avez besoin des indices
    * Si vous utilisez *range*, préciser les indices de début et l'incrément peut clarifier l'algorithme
