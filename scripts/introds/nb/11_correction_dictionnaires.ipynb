## Comptage d'occurrences de mots

À présent nous allons utiliser les dictionnaires pour compter le nombre de mots d'un texte.

### Travail préparatoire
Dans un premier temps, nous préparons le travail en choisissant le texte (le zen de Python) et en utilisant les expressions régulières pour récupérer la liste des mots du texte. C'est à partir de cette liste que vous pourrez utiliser les dictionnaires pour compter le nombre d'occurrences des mots dans le texte. Au début, tout est écrit. Vous n'avez qu'à exécuter les cellules et comprendre.

```python
zen = """The Zen of Python, by Tim Peters

Beautiful is better than ugly.
Explicit is better than implicit.
Simple is better than complex.
Complex is better than complicated.
Flat is better than nested.
Sparse is better than dense.
Readability counts.
Special cases aren't special enough to break the rules.
Although practicality beats purity.
Errors should never pass sile<
ntly.
Unless explicitly silenced.
In the face of ambiguity, refuse the temptation to guess.
There should be one-- and preferably only one --obvious way to do it.
Although that way may not be obvious at first unless you're Dutch.
Now is better than never.
Although never is often better than *right* now.
If the implementation is hard to explain, it's a bad idea.
If the implementation is easy to explain, it may be a good idea.
Namespaces are one honking great idea -- let's do more of those!"""
```

La récupération de la liste des mots est faite en utilisant les expressions régulières. Pour en savoir plus sur les expressions régulières, voir https://docs.python.org/fr/3/howto/regex.html.

```python
# Un petit aperçu des expressions régulières
import re
liste_mots = re.split(r'[\W]+', zen)
print(liste_mots)
```

Observez qu'il y a une chaîne vide en trop dans la liste des mots. Supprimez-la (en écrivant la bonne instruction&nbsp;!

```python
# Plusieurs possibilités : par exemple
#del(liste_mots[len(liste_mots)-1])
liste_mots.remove("")
print(liste_mots)
```

### Calcul du dictionnaire des fréquences

Dans la cellule suivante, calculez un dictionnaire nommé *dico* comptant pour chaque mot son nombre d'occurrences.

```python
# Version 1
dico = {}
for mot in liste_mots:
    if mot in dico:
        dico[mot] += 1
    else:
        dico[mot] = 1
# Version 2 - un peu plus tard dans le noteboook
dico = {}
for mot in liste_mots:
    mot = mot.lower()
    if mot in dico:
        dico[mot] += 1
    else:
        dico[mot] = 1
```

```python
# Affichage du dictionnaire
print(dico)
```

Si on observe bien le dictionnaire, on peut observer que certains mots apparaissent avec des casses de caractères différentes comme par exemple "The" et "the". 
Trouvez la méthode de la classe *String* permettant de transformer tous les caractères en minuscule et modifiez votre construction de dictionnaire précédente pour n'y placer que des mots en minuscule.

### Affichage triée de l'information
L'affichage brut du dictionnaire avec print n'est pas très lisible car l'information n'est pas triée.
Le souci est qu'un dictionnaire n'est pas une information directement triable.
Il est possible d'utiliser la fonction *sorted( )* sur un dictionnaire mais la fonction retourne la liste des mots en ordre alphabétique.

```python
print(sorted(dico))
```

Pour pouvoir trier comme on le souhaite l'information, il est possible de récupérer le dictionnaire sous forme d'une liste de couples grâce à la méthode *items( )*.

```python
print(sorted(dico.items()))
```

L'information est toujours triée par ordre alphabétique des clés, mais il est possible de préciser le critère de tri et même l'ordre de tri (croissant, décroissant).

```python
# Possibilité 1
print(sorted(dico.items(), key=lambda x: x[1], reverse=True))
```

```python
# Possibilité 2
from operator import itemgetter, attrgetter
print(sorted(dico.items(), key=itemgetter(1,0), reverse=True))
```

```python
# Possibilité 3 : avec un tri par ordre décroissant des fréquences
#     mais un tri par ordre croissant des clés de mêmes fréquences
# Dans ce qui suit, on utilise le fait que les tris en Python sont stables
#    (n'hésitez pas à rechercher ce qu'est un tri stable)
# On récupère le dictionnaire sous forme d'une liste triée par ordre 
#    alphabétique des clés
dic_temp = sorted(dico.items())
# Puis on trie la liste de couple résultats par ordre décroissant des fréquences
print(sorted(dic_temp, key=itemgetter(1), reverse=True))
```

Il ne reste plus qu'à écrire un traitement pour afficher les informations sous la forme :<br/>
is : 10<br/>
better : 8<br/>
...

À vous de jouer :

```python
from operator import itemgetter, attrgetter
a_afficher = sorted(dico.items(), key=itemgetter(1,0), reverse=True)
for mot, frequence in a_afficher:
    print(mot, ":", frequence)
```

### La classe defaultdict
Revenons à présent à la construction du dictionnaire, il existe une classe *defaultdict* qui permet d'initialiser les compteurs par défaut à 0 et donc de ne pas avoir à tester si un mot est déjà dans le dictionnaire ou pas. Voir https://docs.python.org/fr/3/library/collections.html#collections.defaultdict.

Reprenez ci-après la création du dictionnaire de comptage

```python
from collections import defaultdict
dico = defaultdict(int)
for mot in liste_mots:
    dico[mot.lower()] += 1
```

```python
print(sorted(dico.items(), key=itemgetter(1,0), reverse=True))
```

La classe *defaultdict* est aussi intéressante pour construire des dictionnaires qui associent à une clé une liste (en remplaçant *int* par *list* dans l'initialisation du dictionnaire).

Utilisez cette possibilité pour définir un dictionnaire inversé associant à chaque fréquence de mots la liste des mots associés. Appelez *dico_inverse* la variable stockant ce nouveau dictionnaire.

```python
pass # construction du dictionnnaire inversé à mettre en oeuvre
dico_inverse = defaultdict(list)
for cle, valeur in dico.items():
    dico_inverse[valeur].append(cle)
```

```python
print(dico_inverse)
```

Dans la cellule suivante, affichez le dictionnaire inversé par ordre décroissant des fréquences, tout en affichant la liste des mots pour une fréquence donnée par ordre alphabétique.

```python
liste_couples = sorted(dico_inverse.items(), reverse=True)
for frequence, liste in liste_couples:
    print(frequence,":",sorted(liste))
```
