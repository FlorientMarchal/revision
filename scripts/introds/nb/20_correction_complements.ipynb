```python
def insertion_dans_liste_triee( lis, val):
    """
    Insère val dans lis en préservant l'ordre
    
    lis : liste de nombres
    val : nombre
    Précondition : lis est triée
    Postcondition : lis est triée (et contient val en plus)
    Remarque : pas de retour (la représentation des listes en mémoire implique la modification)
    """
    lis.append(val) # ajout de val à la fin
    # Descente de val à sa position
    i = len(lis)-2 # on part de la dernière valeur de la liste lis initiale
    pos_trouve = False
    while i >= 0 and not pos_trouve:
        if lis[i] >= val:
            lis[i+1]=lis[i]
            i -= 1
        else:
            pos_trouve = True
    lis[i+1] = val
```

**Exercice** : Écrire une fonction permettant de tester si une liste est un palindrome.

```python
# Approche générique
def est_palindrome( lis ):
    """
    Détermine si lis est un palindrome

    Retourne True si lis est un palindrome, False sinon
    lis : liste
    """
    is_pal = True
    i = 0
    while (i < len(lis)//2) and is_pal:
        if lis[i] == lis[len(lis)-i-1]:
            i += 1
        else:
            is_pal = False
    return is_pal
```

```python
# tests
liste_tests = [
    ([0, 1, 3, 2, 4, 4, 2, 3, 1, 0], True, "Liste palindromique de longueur paire"),
    ([0, 1, 3, 2, 4, 2, 3, 1, 0], True, "Liste palindromique de longueur impaire"),
    ([], True, "Liste vide"),
    ([1], True, "Liste avec un élément"),
    ([0, 1, 3, 2, 4, 4, 2, 3, 1, 2], False, "Liste non palindromique de longueur paire - Pb en position 0"),
    ([0, 1, 3, 2, 4, 1, 2, 3, 1, 0], False, "Liste non palindromique de longueur paire - Pb en position milieu"),
    ([0, 1, 3, 2, 4, 3, 2, -3, 1, 0], False, "Liste non palindromique de longueur paire - Pb en position quelconque"),
    ([0, 1, 3, 2, 4, 2, 3, 1, 1], False, "Liste non palindromique de longueur impaire - Pb en position 0"),
    ([0, 1, 3, 2, 4, -2, 3, 1, 1], False, "Liste non palindromique de longueur impaire - Pb en position milieu"),
    ([0, 1, 3, 2, 4, 2, 3, -1, 1], False, "Liste non palindromique de longueur impaire - Pb en position quelconque"),   
]
test( est_palindrome, liste_tests)
```

```python
# Approche python
def est_palindrome( lis ):
    """
    Détermine si lis est un palindrome

    Retourne True si lis est un palindrome, False sinon
    lis : liste
    """
    return lis == [ elt for elt in reversed(lis)]
```

**Exercice :** Écrire une fonction permettant de déterminer si deux listes contiennent les mêmes valeurs et en même nombre d'occurrences.

```python
# approche générique
def sont_equivalentes( lis1, lis2 ):
    """
    Déterminer si deux listes sont équivalents, c'est-à-dire 
    contiennent les mêmes valeurs et en même nombre d'occurrences.
    
    lis1, lis2 : listes
    """
    if len(lis1) != len(lis2):
        return False
    else:
        lis1bis = sorted(lis1)
        lis2bis = sorted(lis2)
        egales = True
        i = 0
        while i < len(lis1bis) and egales:
            if lis1bis[i]==lis2bis[i]:
                i += 1
            else:
                egales = False
        return egales
```

```python
liste_tests = [
    (([],[]), True, "listes vides"),
    (([1, 1, 1], [1, 1]), False, "mêmes valeurs mais pas le même nombres d'occurrences v1"),
    (([1, 2, 1, 2, 2, 1], [2, 1, 2, 1, 1]), False, "mêmes valeurs mais pas le même nombres d'occurrences v2"),
    (([1, 2, 1, 2, 2, 1], [2, 1, 2, 1, 1, 2]), True, "mêmes valeurs, mêmes nb occurrences en ordre différent"),
]
# N'hésitez pas à en rajouter
test(sont_equivalentes, liste_tests)
```

**Exercice.** Écrivez une fonction permettant de déterminer si toutes les lignes d'une liste de listes sont de même longueur.

```python
# approche générique
def est_tableau2D( lis2D ):
    """
    détermine si la liste de listes lis2D représente un tableau 2D (rectangulaire)
    
    Retourne True si toutes les listes de lis2D ont la même longueur
    
    lis2D : liste de listes
    """
    if len(lis2D) == 0:
        return True
    else:
        ok = True
        val_ref = len(lis2D[0])
        i = 1
        while (i < len(lis2D)) and ok:
            if len(lis2D[i]) == val_ref:
                i += 1
            else:
                ok = False
        return ok
```

**Exercice.** Écrivez une fonction pour calculer la somme d'une liste de listes d'entiers.

```python
# Approche générique
def somme2D( lis2D):
    """
    retourne la somme des éléments de lis2D
    
    lis2D : liste de listes de nombres
    """
    s = 0
    for i in range(0, len(lis2D), 1):
        for j in range(0, len(lis2D[i]), 1):
            s += lis2D[i][j]
    return s
```

```python
liste_tests = [
    ([], 0, "liste vide"),
    ([[1,2,1],[],[0,5,1],[-10]],0,"liste 2D quelconque")
]
test( somme2D, liste_tests)
```

```python
# Approche pythonesque
def somme2D( lis2D):
    """
    retourne la somme des éléments de lis2D
    
    lis2D : liste de listes de nombres
    """
    return sum( [ sum(liste) for liste in lis2D])
```

**Exercice.** Écrivez une fonction déterminant si un entier est présent dans une liste de listes d'entiers.

```python
def est_present( lis2D, valeur):
    """ Détermine si valeur est présente dans lis2D
    
    lis2D : liste de listes
    valeur : type de base quelconque
    Retourne True si lis2D contient valeur, retourne False sinon
    """
    trouve = False
    i = 0 
    while (i < len(lis2D)) and not trouve:
        j = 0
        while (j < len(lis2D[i]) and not trouve):
            if lis2D[i][j] == valeur:
                trouve = True
            else:
                j += 1
        i += 1
    return trouve
```

```python
liste_tests = [
    (([[1,2,1],[],[0,5,1],[-10]], 1), True, "test positif quelconque 1"),
    (([[1,2,1],[],[0,5,1],[-10]], 2), True, "test positif quelconque 2"),
    (([[1,2,1],[],[0,5,1],[-10]], 0), True, "test positif quelconque 0"),
    (([[1,2,1],[],[0,5,1],[-10]], 5), True, "test positif quelconque 5"),
    (([[1,2,1],[],[0,5,1],[-10]], 5), True, "test positif quelconque -10"),
    (([[1,2,1],[],[0,5,1],[-10]], 4), False, "test négatif quelconque 4"),
    (([], 5), False, "test liste vide"),
    (([[],[],[],[]], 5), False, "test liste de listes vides"),
]
test(est_present, liste_tests)
```

```python
def est_present( lis2D, valeur):
    """ Détermine si valeur est présente dans lis2D
    
    lis2D : liste de listes
    valeur : type de base quelconque
    Retourne True si lis2D contient valeur, retourne False sinon
    """
    return any([valeur in liste for liste in lis2D])
```

**Exercice :** Écrire une fonction permettant de déterminer la liste des livres empruntés parmi une liste de livres.

Pour résoudre cet exercice, il faut reprendre la classe Livre dans un notebook précédent. Cela pourrait être fait dans un module. Néanmoins nous recopions cette déclaration dans la cellule suivante.

```python
# Exemple
from enum import Enum
class Etat(Enum):
    LIBRE = 1
    EMPRUNTE = 2

class Livre:
    def __init__( self, numL, titre, auteur, etat):
        self.numL = numL
        self.titre = titre
        self.auteur = auteur
        self.etat = etat
        
    def __str__( self ):
        """ Méthode retournant une description textuelle d'un livre.
            Cette description est utilisée par la fonction print
            """
        return "Numéro = {}, Titre = {}, Auteur = {}, Etat = {}".format(self.numL,
                                                                        self.titre,
                                                                        self.auteur,
                                                                        self.etat)
    
def affichage_biblio( lis_livres):
    """ permet d'afficher une liste de livres """
    for livre in lis_livres:
        print(livre)

def etat( val ):
    """ Retourne l'état correspondant à val """
    if val == "1":
        return Etat.LIBRE
    elif val == "2":
        return Etat.EMPRUNTE
    
def listes2D_vers_liste_Livre( lis2D ):
    """
    Transforme la liste lis2D en une liste de Livre
    
    Donnée : lis2D : liste de listes de chaîne de caractères
    Précondition : toutes les listes de lis2D ont 4 éléments dans l'ordre
        (numéro, titre, auteur, etat)
    Résultat : liste de valeurs de type Livre
    """
    return [ Livre(lis[0], lis[1], lis[2], etat(lis[3])) for lis in lis2D]

from manipulation_csv import lire_fichier_csv
biblio = listes2D_vers_liste_Livre( lire_fichier_csv("bibliotheque.csv"))
affichage_biblio(biblio)
```

```python
# Approche générique
def livres_empruntes( biblio ):
    """ Retourne la liste des livres empruntes
    
    biblio : liste de valeurs de type Livre
    """
    empruntes = []
    for i in range(0, len(biblio), 1):
        if biblio[i].etat == Etat.EMPRUNTE:
            empruntes.append(biblio[i])
    return empruntes
```

```python
# test visuel
empruntes = livres_empruntes( biblio)
affichage_biblio( empruntes )
```

```python
# Approche Python
def livres_empruntes( biblio ):
    """ Retourne la liste des livres empruntes
    
    biblio : liste de valeurs de type Livre
    """
    return [livre for livre in biblio if livre.etat == Etat.EMPRUNTE]
```
