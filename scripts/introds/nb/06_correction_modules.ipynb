## Module d'aide à la mise en place de tests
Lors du développement d'une fonction, il est important de tester son bon fonctionnement. 
Quand une fonction n'est ni une fonction d'affichage, ni une fonction de tests, tester consiste à vérifier que la fonction fournit bien le bon résultat dans des cas connus. Un nombre suffisants de tests (nombre variable selon le problème à résoudre) doivent être réalisés pour pouvoir supposer que les cas particuliers et le cas général ont tous été envisagés. Pour chaque cas envisagé, tester la fonction consiste à exécuter la fonction sur la liste des paramètres du cas considéré et à vérifier que le résultat de la fonction est égal au résultat attendu (qu'il faut donc connaître).

Il est conseillé de conserver les tests réalisés pour pouvoir les relancer rapidement en cas de modification du code. Ainsi il est déconseillé de réaliser des tests par saisie ou d'écraser un test déjà réalisé par un autre.

Il existe pour Python un module (unittest : https://docs.python.org/fr/3.9/library/unittest.html) conçu pour faciliter la mise en oeuvre de tests y compris dans la conception de classes (non abordés à ce stade et dont nous dirons que peu de choses). Nous ne l'abordons pas. Par contre, nous mettons à disposition, une fonction simple pour tester des fonctions que vous écrirez dans le cadre de ce module. Cette fonction est mise à disposition dans le module *aide_test*.

```python
# %load aide_test
"""
Module contenant une fonction pour faciliter les tests

G. Richomme, 27 juin 2021
"""
def test(fonction_a_tester, liste_tests):
    """
    Permet de lancer des tests
    
    Données
    fonction_a_tester : nom d'une fontion
    liste_tests : listes de triplets
        - composante 1 : tuples des paramètres
        - composante 2 : valeur attendue
        - composante 3 : chaîne décrivant le test
    """
    nb_prob = 0
    for data, res_attendu, test in liste_tests:
        if type(data) == tuple:
            res_reel = fonction_a_tester(*data)
        else:
            res_reel = fonction_a_tester(data)
        if res_reel != res_attendu:
            print("------\nprobleme avec test : {}".format(test))
            print("Données = {}".format(data))
            print("Résultat attendu = {}".format(res_attendu))
            print("Résultat réel    = {}".format(res_reel))
            nb_prob += 1
    print("{} problème(s) sur {} tests".format(nb_prob, len(liste_tests)))
```

Le module ne contient qu'une fonction : *test()*. Ci-suit un exemple d'utilisation de cette fonction de test de bon fonctionnement d'une fonction.

```python
def essai(x):
    """ une fonction simple à tester """
    return x == 0

# liste des tests effectués
liste_tests = [ 
    ((0),False,"avec 0"),# Correction de l'erreur : il faut mettre True à la place de False sur cette ligne
    ((1),False,"avec 1"),
]

# lancement du test
test(essai, liste_tests)

# Oups ! Une erreur c'est glissée dans cette cellule... corrigez-la
# Remarque péda : dans la conception du 1er test, la valeur attendue pour le retour doit être True
```
